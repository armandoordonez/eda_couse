**Objetivo.** Revisar la salud general de la tabla y dejarla lista para analizar: nombres de columnas, duplicados, fechas y formatos de números. Al final de este paso se crea una tabla nueva llamada `df_limpio`. La tabla original `df` no se toca.

### 3.1 Nombres de columnas y duplicados

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> Con el DataFrame df: 1) muestra los nombres de las columnas tal como están, 2) cuenta cuántas filas están completamente duplicadas, 3) crea una copia llamada df_limpio sin duplicados y con los nombres de columnas en minúsculas, sin tildes, sin espacios al inicio o al final, con guiones bajos en lugar de espacios y sin símbolos como % o paréntesis.

```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: nombres y duplicados
import unicodedata, re
print("Nombres originales:", list(df.columns))
print("Filas duplicadas:", df.duplicated().sum())

def limpiar_nombre(c):
    c = unicodedata.normalize("NFKD", c.strip()).encode("ascii", "ignore").decode()
    c = re.sub(r"[^0-9a-zA-Z]+", "_", c.lower()).strip("_")
    return c

df_limpio = df.drop_duplicates().copy()
df_limpio.columns = [limpiar_nombre(c) for c in df_limpio.columns]
df_limpio = df_limpio.rename(columns={"satisfaccion_1_5": "satisfaccion"})
print("Nombres nuevos:", list(df_limpio.columns))
print("Filas después de quitar duplicados:", len(df_limpio))
```

### 3.2 Fechas y periodo

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> En df_limpio, la columna fecha tiene formatos mezclados: 2025-03-14, 14/03/2025 y Mar 14 2025, y algunos textos como 'sin fecha'. Convierte la columna a tipo fecha probando los tres formatos, deja vacío lo que no se pueda convertir, y muestra la fecha mínima, la máxima y cuántas ventas hay por mes.

```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: fechas
texto = df_limpio["fecha"].astype(str).str.strip()
f1 = pd.to_datetime(texto, format="%Y-%m-%d", errors="coerce")
f2 = pd.to_datetime(texto, format="%d/%m/%Y", errors="coerce")
f3 = pd.to_datetime(texto, format="%b %d %Y", errors="coerce")
df_limpio["fecha"] = f1.fillna(f2).fillna(f3)
print("Fechas sin convertir:", df_limpio["fecha"].isna().sum())
print("Periodo:", df_limpio["fecha"].min().date(), "a", df_limpio["fecha"].max().date())
df_limpio["mes"] = df_limpio["fecha"].dt.to_period("M").astype(str)
df_limpio["mes"].value_counts().sort_index().plot(kind="bar", title="Ventas por mes", figsize=(10, 3));
```

### 3.3 Números guardados como texto

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> En df_limpio hay números guardados como texto. La columna precio_unitario a veces trae signo $ y puntos de miles (ejemplo: $45.900). La columna descuento a veces trae % (10%), a veces decimal (0.1) y a veces entero (10): conviértela a porcentaje entre 0 y 100. Convierte también cant, valor_total, edad, satisfaccion y dias_entrega a número, dejando vacío lo que no se pueda convertir. Muestra los tipos de dato al final.

```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: números
def a_numero(serie):
    return pd.to_numeric(serie.astype(str).str.replace("$", "", regex=False)
                         .str.replace(".", "", regex=False).str.strip(), errors="coerce")

df_limpio["precio_unitario"] = a_numero(df_limpio["precio_unitario"])

d = df_limpio["descuento"].astype(str).str.replace("%", "", regex=False).str.strip()
d = pd.to_numeric(d, errors="coerce")
df_limpio["descuento"] = np.where(d < 1, d * 100, d)

for col in ["cant", "valor_total", "edad", "satisfaccion", "dias_entrega"]:
    df_limpio[col] = pd.to_numeric(df_limpio[col], errors="coerce")

df_limpio.dtypes
```

### Estandarizar las categorías

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> En df_limpio estandariza estas columnas: ciudad (Cali, Bogotá, Medellín, Barranquilla, Cartagena), canal (Tienda, Online, App), genero (Femenino, Masculino, Otro, y deja vacío 'No responde' o celdas vacías), segmento (Básico, Estándar, Premium), categoria (primera letra en mayúscula) y devolucion (Sí o No; los valores 1 y 0 significan Sí y No). Muestra los valores únicos de cada columna al final para comprobar.

```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: estandarizar categorías
def base(s):
    s = s.fillna("").astype(str).str.strip().str.lower()
    return s.apply(lambda x: unicodedata.normalize("NFKD", x).encode("ascii", "ignore").decode())

c = base(df_limpio["ciudad"])
df_limpio["ciudad"] = np.select(
    [c.str.contains("cali"), c.str.contains("bogota"), c.str.contains("medellin"),
     c.str.contains("barranquilla|b/quilla"), c.str.contains("cartagena")],
    ["Cali", "Bogotá", "Medellín", "Barranquilla", "Cartagena"], default=None)

c = base(df_limpio["canal"])
df_limpio["canal"] = np.select([c.str.contains("tienda"), c.str.contains("online|web"), c.str.contains("app")],
                               ["Tienda", "Online", "App"], default=None)

c = base(df_limpio["genero"])
df_limpio["genero"] = np.select([c.isin(["f", "femenino", "mujer"]), c.isin(["m", "masculino", "hombre"]), c == "otro"],
                                ["Femenino", "Masculino", "Otro"], default=None)

c = base(df_limpio["segmento"])
df_limpio["segmento"] = np.select([c == "basico", c == "estandar", c == "premium"],
                                  ["Básico", "Estándar", "Premium"], default=None)

df_limpio["categoria"] = df_limpio["categoria"].astype(str).str.strip().str.capitalize()

c = base(df_limpio["devolucion"])
df_limpio["devolucion"] = np.select([c.isin(["si", "1"]), c.isin(["no", "0"])], ["Sí", "No"], default=None)

for col in ["ciudad", "canal", "genero", "segmento", "categoria", "devolucion"]:
    print(col, "→", df_limpio[col].value_counts(dropna=False).to_dict())
```

### Outliers y valores imposibles

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> En df_limpio, muestra un resumen estadístico (mínimo, máximo, promedio, mediana) de las columnas continuas. Luego identifica valores imposibles: edad menor a 18 o mayor a 100, cant menor o igual a 0 o mayor a 100, satisfaccion fuera del rango 1 a 5 y dias_entrega mayor a 60.
```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: outliers
display(df_limpio[continuas].describe().T.round(1))

reglas = {
    "edad": (df_limpio["edad"] < 18) | (df_limpio["edad"] > 100),
    "cant": (df_limpio["cant"] <= 0) | (df_limpio["cant"] > 100),
    "satisfaccion": (df_limpio["satisfaccion"] < 1) | (df_limpio["satisfaccion"] > 5),
    "dias_entrega": df_limpio["dias_entrega"] > 60,
}
for col, mascara in reglas.items():
    print(f"{col}: {mascara.sum()} valores imposibles → vacío")
    df_limpio.loc[mascara, col] = np.nan

# valor_total: se recalcula cuando falta o no coincide con cantidad x precio x (1 - descuento)
esperado = df_limpio["cant"] * df_limpio["precio_unitario"] * (1 - df_limpio["descuento"] / 100)
raro = df_limpio["valor_total"].isna() | (df_limpio["valor_total"] > esperado * 2) | (df_limpio["valor_total"] < 0)
print(f"valor_total: {raro.sum()} valores vacíos o inconsistentes → se recalculan")
df_limpio.loc[raro, "valor_total"] = esperado[raro].round(-2)

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
for col, ax in zip(["valor_total", "cant", "edad"], ejes):
    sns.boxplot(x=df_limpio[col], ax=ax, color="#e4eb60")
    ax.set_title(col)
plt.tight_layout()
```

# Paso 5. Análisis bivariado

**Objetivo.** Cruzar **dos variables** para poner a prueba las hipótesis. Aquí empiezan a aparecer las respuestas.

Pistas: categórica contra continua → diagrama de caja o promedio por grupo. Continua contra continua → gráfico de dispersión o correlación. Categórica contra categórica → tabla cruzada con porcentajes.

**Prompt sugerido** (cópielo en Gemini o escriba su propia versión):

> Con df_limpio, quiero probar la hipótesis «a más días de entrega, menor satisfacción». Muestra la satisfacción promedio según los días de entrega en un gráfico de línea, y calcula la correlación entre dias_entrega y satisfaccion. Explica el resultado en lenguaje sencillo.

```python
# Pegue o genere aquí el código de su prompt y ejecútelo
```

```python
#@title Código de respaldo: días de entrega contra satisfacción
envio = df_limpio.dropna(subset=["dias_entrega", "satisfaccion"])
promedio = envio.groupby("dias_entrega")["satisfaccion"].agg(["mean", "count"])
promedio = promedio[promedio["count"] >= 5]
promedio["mean"].plot(marker="o", color="#5454e9", figsize=(9, 4),
                      title="Satisfacción promedio según días de entrega")
plt.ylabel("Satisfacción (1 a 5)"); plt.xlabel("Días de entrega")
print("Correlación:", round(envio["dias_entrega"].corr(envio["satisfaccion"]), 2))
```
