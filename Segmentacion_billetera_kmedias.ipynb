## Análisis de Correlación y RFE

```python
# =============================================================================
# 5. Eliminación de Variables Correlacionadas
# =============================================================================
# Calcular la matriz de correlación
corr_matrix = X_scaled.corr()

# Crear el mapa de calor de la matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Matriz de Correlación - Antes de la Eliminación de Variables")
plt.show()

# Aplicar máscara para eliminar variables altamente correlacionadas
corr_matrix_abs = corr_matrix.abs()
mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
filtered_corr = corr_matrix_abs.where(mask & (corr_matrix_abs > 0.8))
high_corr_features = [column for column in filtered_corr.columns if any(filtered_corr[column] > 0.7)]
X_corr = X_scaled.drop(columns=high_corr_features, errors='ignore')

print("Variables eliminadas por alta correlación:", high_corr_features)
```

```python
# =============================================================================
# 6. Aplicación de PCA o RFE
# =============================================================================
# Importar RFE
from sklearn.feature_selection import RFE

use_pca = False
use_rfe = True

if use_pca:
    pca = PCA(n_components=0.8)  # Retener el 80% de la varianza explicada
    X_pca = pca.fit_transform(X_corr)

    # Número de componentes seleccionados
    print(f"\nNúmero de componentes principales seleccionados: {pca.n_components_}")

    # Frecuencia acumulada de varianza explicada
    explained_variance_ratio = np.cumsum(pca.explained_variance_ratio_)

    plt.figure(figsize=(8, 5))
    plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, marker='o', linestyle='--')
    plt.xlabel('Número de Componentes Principales')
    plt.ylabel('Varianza Explicada Acumulada')
    plt.title('Frecuencia Acumulada de Varianza Explicada por PCA')
    plt.grid()
    plt.show()

    # Convertir a DataFrame para facilitar la visualización
    X_for_clustering = pd.DataFrame(X_pca, index=X_corr.index)

    # Graficar PC1 vs PC2, PC1 vs PC3, y PC2 vs PC3
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    axes[0].scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.5)
    axes[0].set_xlabel('PC1')
    axes[0].set_ylabel('PC2')
    axes[0].set_title('PC1 vs PC2')

    axes[1].scatter(X_pca[:, 0], X_pca[:, 2], alpha=0.5)
    axes[1].set_xlabel('PC1')
    axes[1].set_ylabel('PC3')
    axes[1].set_title('PC1 vs PC3')

    axes[2].scatter(X_pca[:, 1], X_pca[:, 2], alpha=0.5)
    axes[2].set_xlabel('PC2')
    axes[2].set_ylabel('PC3')
    axes[2].set_title('PC2 vs PC3')

    plt.show()

    # Visualización 3D de PC1, PC2 y PC3
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], alpha=0.5)

    ax.set_xlabel('PC1')
    ax.set_ylabel('PC2')
    ax.set_zlabel('PC3')
    ax.set_title('Visualización 3D de PCA')

    plt.show()

elif use_rfe:
    # Definir la variable objetivo y las características
    y = X_corr["Nivel_Interaccion_Billetera"]
    X_rfe_input = X_corr.drop(columns=["Nivel_Interaccion_Billetera"], errors='ignore')

    # Aplicar RFE con Random Forest
    rfe_selector = RFE(
        estimator=RandomForestRegressor(n_estimators=100, random_state=42),
        n_features_to_select=5,  # Selecciona cuántas características deseas
        step=1
    )
    rfe_selector.fit(X_rfe_input, y)

    # Obtener las columnas seleccionadas
    selected_cols = X_rfe_input.columns[rfe_selector.support_]
    print("Features seleccionadas con RFE:\n", selected_cols.tolist())

    # Definir el DataFrame final con las variables seleccionadas
    X_for_clustering = X_rfe_input[selected_cols]

    # Obtener la importancia de cada variable seleccionada
    feature_importance = pd.DataFrame({
        'Feature': selected_cols,
        'Importance': rfe_selector.estimator_.feature_importances_
    }).sort_values(by="Importance", ascending=False)

    # Mostrar la importancia de las variables en un gráfico
    plt.figure(figsize=(10, 6))
    sns.barplot(x=feature_importance["Importance"], y=feature_importance["Feature"], palette="viridis")
    plt.xlabel("Importancia de la Variable")
    plt.ylabel("Variable")
    plt.title("Importancia de Predicción de las Variables Seleccionadas (RFE + Random Forest)")
    plt.show()

else:
    X_for_clustering = X_scaled.copy()  # Usar todas las disponibles en X_scaled
```

## Análisis de Silhouette

```python
# =============================================================================
# Análisis Silhouette Detallado (Gráfico de Silhouette)
# =============================================================================

from sklearn.metrics import silhouette_score, silhouette_samples
import numpy as np

# Etiquetas y datos
labels_cluster = model.labels_
X_cluster = X_for_clustering

# 1. Silhouette Score global
silhouette_avg = silhouette_score(X_cluster, labels_cluster)
print(f"✅ Silhouette Score global: {silhouette_avg:.4f}")

# 2. Interpretación automática
print(f"\n🔍 Interpretación del Silhouette Score ({silhouette_avg:.4f}):")
if silhouette_avg < 0.2:
    print("⚠️ Estructura débil → Los grupos están poco definidos o se solapan.")
elif silhouette_avg < 0.4:
    print("🟡 Agrupamientos razonables → Se pueden mejorar, pero hay cierta separación.")
elif silhouette_avg < 0.6:
    print("🟢 Buena separación → Los clústeres tienen coherencia y compacidad.")
else:
    print("✅ Muy buenos clústeres → Excelente definición entre los grupos.")

# 3. Valores individuales
silhouette_vals = silhouette_samples(X_cluster, labels_cluster)

# 4. Gráfico de Silhouette por clúster
n_clusters = len(np.unique(labels_cluster))
y_lower = 10

plt.figure(figsize=(10, 6))
for i in range(n_clusters):
    ith_vals = silhouette_vals[labels_cluster == i]
    ith_vals.sort()
    size_i = ith_vals.shape[0]
    y_upper = y_lower + size_i

    color = plt.cm.nipy_spectral(float(i) / n_clusters)
    plt.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_vals, facecolor=color, edgecolor=color, alpha=0.7)
    plt.text(-0.05, y_lower + 0.5 * size_i, str(i))
    y_lower = y_upper + 10

plt.axvline(x=silhouette_avg, color="red", linestyle="--", label=f"Media = {silhouette_avg:.2f}")
plt.title("Gráfico de Silhouette por Clúster", fontsize=13)
plt.xlabel("Valor de Silhouette")
plt.ylabel("Clúster")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Seleccionar solo columnas numéricas (excluyendo la columna 'cluster' si está)
columns_to_fill = df_con_cluster.select_dtypes(include='number').drop(columns=["cluster"], errors='ignore').columns

# Loop por cada variable numérica
for column in columns_to_fill:
    print(f"📊 Estadísticas para la variable: {column}")
    print("-" * 50)

    # Agrupar por clúster y calcular estadísticas
    summary = df_con_cluster.groupby('cluster')[column].agg(['count','mean', 'median', 'std']).round(2).reset_index()

    # Mostrar
    print(summary)
    print("\n")
```

```python
# Seleccionar solo columnas categóricas (excluyendo 'cluster' si está)
categorical_columns = df_con_cluster.select_dtypes(include='object').columns

# Loop por cada variable categórica
for column in categorical_columns:
    print(f"📋 Estadísticas para la variable categórica: {column}")
    print("-" * 50)

    # Calcular la moda por clúster
    mode_per_cluster = df_con_cluster.groupby("cluster")[column].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan).reset_index()

    # Mostrar la moda por clúster
    print("Moda por clúster:")
    print(mode_per_cluster)

    print("\n")
```
